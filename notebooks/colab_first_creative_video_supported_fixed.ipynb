# First Complete Creative Video — Supported (Fixed)

This lightweight notebook delegates heavy work to scripts. Use this if the main supported notebook fails.

```python
# Configuration
BRANCH = 'tga892338-rgb-refactor-colab-ready'
REPO_URL = 'https://github.com/tga892338-rgb/openmontage-colab.git'
PROJECT_DIR = 'openmontage-colab'
TTS_BACKEND = 'qwen'  # or 'chatterbox'
DRY_RUN = True  # set False for real run
IDEA = 'What would happen if Earth suddenly stopped rotating?'
print('Configured TTS:', TTS_BACKEND, 'DRY_RUN:', DRY_RUN)
```

```python
# Clone the repo if needed and cd into it
import os, subprocess, sys, shlex
if not os.path.isdir(PROJECT_DIR):
    subprocess.check_call(['git','clone','-b',BRANCH,REPO_URL,PROJECT_DIR])
os.chdir(PROJECT_DIR)
print('CWD:', os.getcwd())
```

```python
# Generate the shot plan (director)
import subprocess, sys, shlex
cmd = [sys.executable, 'tools/director_first_creative.py', '--idea', IDEA]
print('Running:', ' '.join(shlex.quote(p) for p in cmd))
subprocess.check_call(cmd)
```

```python
# Run orchestration (captured) and display final video
import subprocess, sys, os, shlex, traceback
from IPython.display import Video, display
PROJECT_PATH = 'projects/first-creative-video'
cmd = [sys.executable, 'scripts/run_orchestration_capture.py', '--project', PROJECT_PATH, '--tts', TTS_BACKEND]
if DRY_RUN:
    cmd.append('--dry-run')
print('Running orchestration (captured):', ' '.join(shlex.quote(p) for p in cmd))
try:
    subprocess.check_call(cmd)
except subprocess.CalledProcessError as e:
    print('Orchestration failed with return code', e.returncode)
    log = os.path.join(PROJECT_PATH, 'orchestration.log')
    if os.path.exists(log):
        print('
--- Last 400 lines of orchestration.log ---')
        with open(log, 'r', encoding='utf-8', errors='replace') as f:
            print(''.join(f.readlines()[-400:]))
    else:
        traceback.print_exc()
    raise
final = os.path.join(PROJECT_PATH, 'final.mp4')
if os.path.exists(final):
    print('Final video:', final)
    display(Video(final, embed=True, width=640))
else:
    print('Final video not found at', final)
```
